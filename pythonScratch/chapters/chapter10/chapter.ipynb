```python
# 访客　编写一个程序，提示用户输入其名字。在用户做出响应后，将其名字写入文件 guest.txt。
input_text = input("Please input your name: ")
path = Path("guest.txt")
path.write_text(input_text)
```

```python
# 访客簿　编写一个 while 循环，提示用户输入其名字。
# 收集用户输入的所有名字，将其写入 guest_book.txt，并确保这个文件中的每条记录都独占一行。
a = 5
user_names = ""
while a > 0:
    user_name = input("Please input your name:")
    user_names += f"{user_name}\n"
    path = Path("guest_book.txt")
    path.write_text(user_names)
    a = a-1
```

```python
path = Path("alice.txt")
try:
    contents = path.read_text(encoding='utf-8')
except FileNotFoundError:
    print(f"Sorry, the file {path} is not exists.")
else:
    lines = contents.split()
    num_words = len(lines)
    print(f"This file {path} has {num_words} words.")
```

```python
def count_words(path):
    try:
        contents = path.read_text(encoding='utf-8')
    except FileNotFoundError:
        print(f"Sorry, the file {path} is not exists.")
    else:
        lines = contents.split()
        num_words = len(lines)
        print(f"This file {path} has {num_words} words.")
```
