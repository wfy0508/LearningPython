## 8 函数

```python
# 传递任意数量的实参
# 形参名*toppings中的星号让Python创建一个名为toppings的元组，该元组包含函数收到的所有值。
def make_pizza(*toppings):
    print(toppings)
make_pizza('apple')
```

```python
# Python会将实参封装到一个元组中，即便函数只收到一个值也是如此
make_pizza(['apple', 'pear', 'orange'])
make_pizza('apple', 'pear', 'orange')
```

```python
def make_pizza(*toppings):
    for topping in toppings:
        print(f"{topping} pizza.")

make_pizza('apple', 'pear', 'orange')
# 传入列表，会被认为是一个参数
make_pizza(['apple', 'pear', 'orange'])
```

```python
# 如果要让函数接受不同类型的实参，必须在函数定义中将接纳任意数量实参的形参放在最后。
def make_pizza(size, *toppings):
    for topping in toppings:
        print(f" - Make a {size} inch {topping} pizza.")

make_pizza(12, 'apple', 'pear', 'orange')
```

```python
def build_profile(first_name, last_name, **user_info):
    user_info['first_name'] = first_name
    user_info['last_name'] = last_name
    print(user_info)


build_profile('albert', 'einstein', location='princeton', field='pyhsics')
```
